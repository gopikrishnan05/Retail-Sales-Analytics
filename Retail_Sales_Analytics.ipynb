Age Analysis

Does spending differ across age groups?

```python
def age_group(age):
    if 20 <= age <= 29:
        return ('20 - 29')
    elif 30 <= age <= 39:
            return ('30 - 39')
    elif 40 <= age <= 49:
            return ('40 - 49')
    elif 50 <= age <= 59:
            return ('50 - 59')
    elif 60 <= age <= 70:
                return ('60 - 70')

df['Age_Group'] = df['Age'].apply(age_group)


    
```
